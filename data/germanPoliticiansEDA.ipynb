```python
import pandas as pd

df = pd.read_excel("Deutsche Politik auf Tik Tok - Eine Sammlung von Martin Fuchs @wahl_beobachter.xlsx", skiprows=1)

df.head()
```

## Make a column with only the profile name

```python
df = df[~df["Tik Tok-Profil"].isna()]#only keep the rows where we have profiles

def extract_name(profile_url:str) -> str:
    name = profile_url.split("@")[-1].rstrip()
    return name

df["profile_name"] = df["Tik Tok-Profil"].apply(extract_name)

df.head(50)
```
